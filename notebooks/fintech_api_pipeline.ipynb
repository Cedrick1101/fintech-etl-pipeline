### Define the API endpoint. The API configuration section

```python
BASE_CURRENCY = "USD"

API_URL = (
    f"https://open.er-api.com/v6/latest/{BASE_CURRENCY}"
)

print(API_URL) # Print the API URL to verify it's correct
```

### Extraction 

```python
response = requests.get(API_URL, timeout=30)
print("status_code: " , response.status_code)

response.raise_for_status() # Raise an error if the request was unsuccessful
print("Request Successful")

api_data = response.json() # Convert the response to JSON format

print(api_data["result"]) # 

print("CURRENCY BASE CODE: " +api_data["base_code"])

               
```
